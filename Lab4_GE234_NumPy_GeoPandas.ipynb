## 🔹 ตัวอย่างที่ 1: ใช้ NumPy คำนวณข้อมูล NDVI จากภาพ Raster


```python

import numpy as np

# สร้างข้อมูลตัวอย่างสำหรับภาพ NDVI (Normalized Difference Vegetation Index)
nir = np.array([[0.7, 0.8, 0.6], [0.9, 0.5, 0.3], [0.4, 0.7, 0.2]])
red = np.array([[0.3, 0.4, 0.2], [0.5, 0.3, 0.1], [0.2, 0.3, 0.1]])

# คำนวณค่า NDVI
ndvi = (nir - red) / (nir + red)

print("ค่า NDVI:")
print(ndvi)
```


## 🔹 ตัวอย่างที่ 2: ใช้ NumPy คำนวณค่าเฉลี่ยและค่ามากสุดของ NDVI


```python

print(f"ค่าเฉลี่ย NDVI: {np.mean(ndvi):.2f}")
print(f"ค่า NDVI สูงสุด: {np.max(ndvi):.2f}")
print(f"ค่า NDVI ต่ำสุด: {np.min(ndvi):.2f}")
```


## 🔹 ตัวอย่างที่ 3: ใช้ GeoPandas โหลดและวิเคราะห์ข้อมูล Shapefile


```python
import geopandas as gpd

# โหลดข้อมูลชั้นข้อมูลจังหวัดของประเทศไทย
gdf = gpd.read_file("/content/tha_admbnda_adm1_rtsd_20190221.gpkg")

# แสดงข้อมูล 5 แถวแรก
print(gdf.head())
```


## 🔹 ตัวอย่างที่ 4: คำนวณพื้นที่ของแต่ละจังหวัด


```python

# ตรวจสอบค่า CRS (Coordinate Reference System)
print(gdf.crs)

# คำนวณพื้นที่ของแต่ละจังหวัด (หน่วยเป็นตารางเมตร)
gdf["area_sqkm"] = gdf.geometry.area / 1e6

# แสดงพื้นที่จังหวัด 5 อันดับแรกที่ใหญ่ที่สุด
print(gdf.nlargest(5, "area_sqkm")[["ADM1_TH", "area_sqkm"]])
```


## 🔹 ตัวอย่างที่ 5: ใช้ GeoPandas ทำ Spatial Join


```python
# โหลดข้อมูลชั้นข้อมูลอำเภอ
gdf_districts = gpd.read_file("/content/tha_admbnda_adm1_rtsd_20190221.gpkg")

# ทำ Spatial Join ระหว่างอำเภอกับจังหวัด
gdf_joined = gpd.sjoin(gdf_districts, gdf, how="inner", predicate="within")

# แสดงตัวอย่างข้อมูลที่เชื่อมโยงกัน
print(gdf_joined.head())
```

**แบบฝึกหัด 1**: ใช้ NumPy คำนวณค่า Mean, Max, Min ของค่า NDVI ในอาร์เรย์ที่สร้างขึ้นเอง

```python
import numpy as np

# 1. สร้างอาเรย์รายชื่อ 7 จังหวัด (String Array)
provinces_7 = np.array([
    "เชียงใหม่", "ขอนแก่น", "ชลบุรี", "นครราชสีมา",
    "ภูเก็ต", "สงขลา", "กาญจนบุรี"
])

# 2. สร้างอาเรย์ค่า NDVI จำลองสำหรับ 7 จังหวัดนี้ (Float Array)
# สมมติค่า NDVI อยู่ในช่วง 0.1 ถึง 0.9
ndvi_values = np.array([0.75, 0.42, 0.38, 0.55, 0.68, 0.49, 0.82])

# 3. คำนวณค่าทางสถิติของ 7 จังหวัดนี้
mean_ndvi = np.mean(ndvi_values)
max_ndvi = np.max(ndvi_values)
min_ndvi = np.min(ndvi_values)

# 4. แสดงผล
print(f"รายชื่อจังหวัด: {provinces_7}")
print(f"ค่า NDVI เฉลี่ยรวม 7 จังหวัด: {mean_ndvi:.2f}")
print(f"ค่าสูงสุด (Max): {max_ndvi:.2f} (จังหวัด {provinces_7[np.argmax(ndvi_values)]})")
print(f"ค่าต่ำสุด (Min): {min_ndvi:.2f} (จังหวัด {provinces_7[np.argmin(ndvi_values)]})")
```

**แบบฝึกหัด 2**: ใช้ GeoPandas โหลด Shapefile ของจังหวัด และคำนวณพื้นที่ของแต่ละจังหวัด

```python
import geopandas as gpd

# 1. โหลดข้อมูล (เปลี่ยนชื่อไฟล์ตามที่คุณโหลดมา)
gdf = gpd.read_file('/content/gadm41_THA_1.json.zip')

# 2. กรองข้อมูลเฉพาะ 7 จังหวัด (ชื่อใน GADM มักเป็นภาษาอังกฤษ)
provinces = ['Chiang Mai', 'Khon Kaen', 'Chon Buri', 'Nakhon Ratchasima',
             'Phuket', 'Songkhla', 'Kanchanaburi']
gdf_7 = gdf[gdf['NAME_1'].isin(provinces)].copy()

# 3. แปลงพิกัดเป็น UTM Zone 47 (EPSG:32647) เพื่อคำนวณพื้นที่เป็นเมตร
gdf_7 = gdf_7.to_crs(epsg=32647)

# 4. คำนวณพื้นที่เป็นตารางกิโลเมตร
gdf_7['area_km2'] = gdf_7.geometry.area / 10**6

# 5. แสดงผล 7 แถวแรก
print("--- ข้อมูลพื้นที่ 7 จังหวัด ---")
print(gdf_7[['NAME_1', 'area_km2']].head(7))

# 6. หาค่าทางสถิติของพื้นที่
import numpy as np
print(f"\nพื้นที่เฉลี่ยของ 7 จังหวัด: {np.mean(gdf_7['area_km2']):.2f} ตร.กม.")
print(f"จังหวัดที่ใหญ่ที่สุดในกลุ่มนี้คือ: {gdf_7.loc[gdf_7['area_km2'].idxmax(), 'NAME_1']}")
```

**แบบฝึกหัด 3**: ใช้ GeoPandas ทำ Spatial Join ระหว่างข้อมูลจังหวัดและอำเภอ

```python
import geopandas as gpd

# 1. โหลดข้อมูล (สมมติใช้ไฟล์จาก GADM)
# เลเยอร์จังหวัด (Polygon แม่)
gdf_prov = gpd.read_file('/content/gadm41_THA_1.json.zip')

# เลเยอร์อำเภอ (Polygon ลูก)
gdf_amphoe = gpd.read_file('/content/gadm41_THA_2.json.zip')

# 2. ตรวจสอบและตั้งค่า CRS ให้ตรงกัน (สำคัญมาก!)
# ทั้งสองไฟล์ต้องใช้ระบบพิกัดเดียวกันถึงจะ Join กันได้
gdf_amphoe = gdf_amphoe.to_crs(gdf_prov.crs)

# 3. ทำ Spatial Join (sjoin)
# เราจะเอา "อำเภอ" เป็นตัวตั้ง แล้วไปดูว่า "อยู่ใน" (within) จังหวัดไหน
joined_gdf = gpd.sjoin(gdf_amphoe, gdf_prov[['NAME_1', 'geometry']], how="left", predicate="within")

# 4. แสดงผล 5 แถวแรกที่มีทั้งชื่ออำเภอ (NAME_2) และชื่อจังหวัด (NAME_1_right)
print("--- ผลลัพธ์การ Join ข้อมูลอำเภอและจังหวัด ---")
print(joined_gdf[['NAME_2', 'NAME_1_right']].head(5))

# 5. นับจำนวนอำเภอในแต่ละจังหวัด (Bonus)
amphoe_count = joined_gdf.groupby('NAME_1_right').size().reset_index(name='count_amphoe')
print("\nจำนวนอำเภอแยกตามจังหวัด (5 แถวแรก):")
print(amphoe_count.head(5))
```

**แบบฝึกหัด 4**: ใช้ NumPy และ GeoPandas ร่วมกันเพื่อหาข้อมูลจังหวัดที่มี NDVI เฉลี่ยสูงสุด

```python
import geopandas as gpd
import numpy as np

# 1. โหลดข้อมูลจังหวัด (ใช้ไฟล์ระดับ 1 จาก GADM ที่แนะนำไป)
gdf = gpd.read_file('/content/gadm41_THA_1.json.zip')

# 2. จำลองค่า NDVI ให้กับทุกจังหวัดในตาราง (สมมติค่าระหว่าง 0.1 - 0.9)
# เราใช้ np.random เพื่อสร้างอาเรย์ตัวเลขที่มีจำนวนเท่ากับจำนวนแถวของจังหวัด
np.random.seed(42) # ล็อกค่าสุ่มให้ได้ผลเหมือนเดิมทุกครั้งที่รัน
gdf['ndvi_value'] = np.random.uniform(0.1, 0.9, size=len(gdf))

# 3. ใช้ NumPy หาค่าสูงสุด และตำแหน่ง (Index) ของค่าสูงสุดนั้น
ndvi_array = gdf['ndvi_value'].to_numpy() # แปลงคอลัมน์เป็น NumPy Array
max_ndvi = np.max(ndvi_array)             # หาค่าสูงสุด
max_index = np.argmax(ndvi_array)         # หาตำแหน่ง Index ที่มีค่าสูงสุด

# 4. ดึงข้อมูลจังหวัดจาก Index ที่หาได้
top_province = gdf.iloc[max_index]

# 5. แสดงผลลัพธ์
print(f"--- ผลการวิเคราะห์ NDVI รายจังหวัด ---")
print(f"จังหวัดที่มีค่า NDVI เฉลี่ยสูงสุดคือ: {top_province['NAME_1']}")
print(f"ค่า NDVI ที่คำนวณได้: {max_ndvi:.4f}")

# Bonus: แสดง 5 อันดับแรกที่มีค่า NDVI สูงสุด
print("\n--- 5 อันดับจังหวัดพื้นที่สีเขียวสูงสุด ---")
print(gdf[['NAME_1', 'ndvi_value']].sort_values(by='ndvi_value', ascending=False).head(5))
```
